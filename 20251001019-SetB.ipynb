````python
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)


def combinations(n, r):
    if n < 0 or r < 0:
        print("Error: n and r must not be negative.")
        return None

    if r > n:
        print("Error: r cannot be greater than n.")
        return None

    return factorial(n) // (factorial(r) * factorial(n - r))


result = combinations(7, 2)
if result is not None:
    print("Number of possible catalyst sets:", result)

combinations(4, 6)
````

````python
def fib_iter(n):
    if n <= 0:
        return []

    fibonacci = [0]

    if n == 1:
        return fibonacci

    fibonacci.append(1)

    while len(fibonacci) < n:
        fibonacci.append(fibonacci[-1] + fibonacci[-2])

    return fibonacci


def fib_rec(k):
    if k < 0:
        return None
    if k == 0:
        return 0
    if k == 1:
        return 1

    return fib_rec(k - 1) + fib_rec(k - 2)


n = 9
iterative_values = fib_iter(n)
recursive_values = [fib_rec(k) for k in range(n)]

print("Both functions agree:", iterative_values == recursive_values)

print(f"{'Sample No.':>12} {'Fibonacci Tag':>15} {'Sampling Time (min)':>22}")

for sample_no, fibonacci_tag in enumerate(iterative_values, start=1):
    sampling_time = fibonacci_tag * 5
    print(f"{sample_no:>12} {fibonacci_tag:>15} {sampling_time:>22}")
````

## 5. Activity 2 – Chemical Equilibrium Problem

Consider the simplified equilibrium reaction:

```text
A + B ⇌ C
```

Assume initially:

- `[A]₀ = 1.0 mol/L`
- `[B]₀ = 1.0 mol/L`
- `[C]₀ = 0 mol/L`

Let `x` be the equilibrium concentration formed of C.

At equilibrium:

```text
[A] = 1 - x
[B] = 1 - x
[C] = x
```

For this simplified example, take:

```text
Kc = [C] / ([A][B])
```

Therefore:

```text
Kc = x / (1 - x)²
```

Rearranging gives:

```text
Kc x² - (2Kc + 1)x + Kc = 0
```

**This is the key chemical-engineering connection of this practical:** the equilibrium expression has been converted into a quadratic equation in `x`.

````python
# plot the measured values of pressure vs. volume and the reference curve P = k/V, where k is the mean of P × V for the measured data. Add a title, label both axes with units, display a legend, add a grid, and use different colors for the measured values and the reference curve. Use a dashed line for the reference curve.
import numpy as np
import matplotlib.pyplot as plt 


volume = np.array([1, 2, 3, 4, 5])
pressure = np.array([3.03, 1.49, 1.01, 0.76, 0.59])

# Calculate k = P × V for each point, then find its mean
k = np.mean(pressure * volume)
print(f"k = {k:.3f} bar·L")

# Generate the reference curve
reference_volume = np.linspace(volume.min(), volume.max(), 200)
reference_pressure = k / reference_volume

# Plot measured values and the reference curve
plt.figure(figsize=(7, 5))
plt.scatter(volume, pressure, color="blue", label="Measured data")
plt.plot(
    reference_volume,
    reference_pressure,
    color="red",
    linestyle="--",
    label=f"P = k/V, k = {k:.3f} bar·L"
)

plt.title("Pressure vs. Volume at Constant Temperature")
plt.xlabel("Volume (L)")
plt.ylabel("Pressure (bar)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
````
