```python
#If elif else statement for marks of the student
# Above 90 - A
# 85 and above - B
# 75 and above - C
marks = int(input("Enter your marks: "))
if marks > 90:
    print("Grade: A")
elif marks >= 85:
    print("Grade: B")
elif marks >= 75:
    print("Grade: C")
else:
    print("Grade: D")
```
